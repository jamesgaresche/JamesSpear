```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x,A,w):
    return A*np.sin(w*x)

x = np.linspace(0,2*np.pi,1000)

Amp = [0.5, 1, 1.5, 2, 2.5]
Freq = [1, 2, 3, 4, 5]

fig, axes = plt.subplots(1, 3, figsize=(25, 5))
for A, w in zip(Amp, Freq):
    axes[0].plot(x, make_sine_wave(x, A, 1),label=f"A = {A}")
    axes[1].plot(x, make_sine_wave(x, 1, w),label=f"w = {w}")
    axes[2].plot(x, make_sine_wave(x, A, w),label=f"A = {A}, w = {w}")

axes[0].set_title("Varying Amplitude", fontsize=16)
axes[0].set_xlabel("x")
axes[0].set_ylabel("y")
axes[0].legend()

axes[1].set_title("Varying Frequency", fontsize=16)
axes[1].set_xlabel("x")
axes[1].set_ylabel("y")
axes[1].legend()

axes[2].set_title("Varying Amplitude and Frequency", fontsize=16)
axes[2].set_xlabel("x")
axes[2].set_ylabel("y")
axes[2].legend()
plt.show()
```
