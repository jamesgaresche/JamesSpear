```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt",skiprows=1)
print(data.shape) #2642 rows and 2 columns

magnitude = data[:,0]
print(np.min(magnitude))
print(np.argmin(magnitude)) # The brightest star is number 1988
print(np.max(magnitude))
print(np.argmax(magnitude)) # The dimmest star is number 264
temperature = data[:,1]
print(np.min(temperature))
print(np.argmin(temperature))
print(np.max(temperature))
print(np.argmax(temperature))
```

```python
log_mag = np.log(magnitude)
log_temp = np.log(temperature)

plt.figure(figsize = (8,6))
plt.scatter(log_temp,log_mag, label = "stars")
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.scatter(log_temp[1988],log_mag[1988], color = "black", s = 300, marker = "*", label = "brightest")
plt.scatter(log_temp[264],log_mag[264], color = "purple", s = 300, marker = "*", label = "dimmest")
plt.scatter(log_temp[2613],log_mag[2613], color = "green", s = 300, marker = "*", label = "coldest")
plt.scatter(log_temp[1923],log_mag[1923], color = "orange", s = 300, marker = "*", label = "hottest")
plt.legend()
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.show()
```
