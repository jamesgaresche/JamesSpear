```python
arr = np.array([
[2, 3, 5],
[14, 6, 8],
[11, 13, 17],
[7, 10, 13]])

def prime(n):
    if n < 2:
        return False
    else:
        for i in range(2,n):
            if n % i == 0:
                return False
            else:
                return True

def prime_row(arr):
    ret = []
    for row in arr:
        if any(prime(n) for n in row):
            ret.append(row)
    return np.array(ret)

print(prime_row(arr))
```

```python
def checkerboard():
    board = np.zeros((8, 8), dtype=int)
    board[::2, ::2] = 1
    board[1::2, 1::2] = 1
    return board

print(checkerboard())

def rev_checkerboard():
    board = np.zeros((8, 8), dtype=int)
    board[1::2, ::2] = 1
    board[::2, 1::2] = 1
    return board

print(rev_checkerboard())
```

```python
universe = np.array(['galaxy', 'clusters'])

def expansion(universe, n):
    ret = []
    for word in universe:
        ret.append((' ' * n).join(word))
    return np.array(ret)

print(expansion(universe, 1))
print(expansion(universe, 2))
```

```python
stars = np.array(np.random.randint(500, 2000, (5, 5))])

def secondDimmest(stars):
    return np.sort(stars, axis=1)[:, -2]

print(secondDimmest(stars))
```
