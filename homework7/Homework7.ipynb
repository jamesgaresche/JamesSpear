```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("state_data.csv")
print(df.shape) # 52 rows and 4 columns
print(df.head()) 
print(df.tail())
''' The different columns are the state abbreviations, latitude coordinates, longitude coordinates, and the states name.
The first state is Alaska and the last is Wyoming as they are in alphabetical order. There are 52 states total because peutro rico and DC
are included '''
```

```python
Lat_med = df['latitude'].median()
Long_med = df['longitude'].median()
print(Lat_med)
print(Long_med)

Lat_dist = np.abs(df['latitude'] - Lat_med)
Long_dist = np.abs(df['latitude'] - Long_med)

df['Lat_dist'] = Lat_dist
df['Long_dist'] = Long_dist

print(df.head())
df.shape # 52 rows and 6 columns



print(f"{df.iloc[df['Lat_dist'].idxmin(),0]} is closest to median latitude. It is off by {df.iloc[df['Lat_dist'].idxmin(),4]}.")
print(f"{df.iloc[df['Lat_dist'].idxmax(),0]} is furthest from median latitude. It is off by {df.iloc[df['Lat_dist'].idxmax(),4]}.")
print(f"{df.iloc[df['Long_dist'].idxmin(),0]} is closest to median latitude. It is off by {df.iloc[df['Long_dist'].idxmin(),5]}.")
print(f"{df.iloc[df['Long_dist'].idxmax(),0]} is closest to median latitude. It is off by {df.iloc[df['Long_dist'].idxmax(),5]}.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('diamonds')
df = df.dropna(axis=1)

print(df.shape) # 53940 rows and 10 columns
print(df.head())
print(df.tail())
''' The first column is how many carats the diamond is, the second is the quality of the cut, the thrid is the color, the foruth the clarity, the fifth the depth and the sixth the table which are both measurements, the seventh, eight, and ninth columns are al x,y,z measurements'''
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

sns.scatterplot(data=df, x="carat", y="price", color="blue", ax=axes[0, 0])
axes[0, 0].set_title("Carats vs Price")
axes[0, 0].set_xlabel("Carats")
axes[0, 0].set_ylabel("Price (Dollars)")

sns.scatterplot(data=df, x="table", y="depth", color="red", ax=axes[0, 1], label = 'data')
axes[0, 1].set_title("Diamond Proportions")
axes[0, 1].set_xlabel("Table (mm)")
axes[0, 1].set_ylabel("Depth (mm)")

sns.histplot(data=df, x="carat", color="green", ax=axes[1, 0])
axes[1, 0].set_title("Distribution of Carats")
axes[1, 0].set_xlabel("Carats")
axes[1, 0].set_ylabel("Frequency")

sns.boxplot(data=df, x="price", color="orange", ax=axes[1, 1])
axes[1, 1].set_title("Price Distribution")
axes[1, 1].set_xlabel("Price (Dollars)")
plt.show()

''' The first plot comparing price and carats shows how the price increases much faster than linearly as carats increase, though there are some linear outliers.
The second plot comparing diamond depth and table shows how the two main measurements determining the propotions and light reflection relate, demonstarting the spread of acceptable cuts with outliers.
The third is a simple histogram of carats showing the distribution of diamond sizes. Lastly, the fourth is a distribtuion of prices, which shows that more than half are below 2500 but there are many outliers and a much larger spread on the upper end.
'''
```
